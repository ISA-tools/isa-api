```python
isa_json = isatab2json.convert(data_dir,use_new_parser=True)
          
with open(os.path.join(data_dir, 'isa.json'), 'w') as out_fp:
            json.dump(isa_json, out_fp)
```

```python
from isatools.convert.json2jsonld import ISALDSerializer
from json import load
```

```python
instance_path = os.path.join(data_dir, "isa.json")

with open(instance_path, 'r') as instance_file:
        instance = load(instance_file)
        instance_file.close()
```

```python

ontology = "wd"

serializer = ISALDSerializer(instance)
serializer.set_ontology(ontology)
serializer.set_instance(instance)
serializer.combined=True
jsonldcontent = serializer.output
```

```python
with open(os.path.join(data_dir, 'isa-rdf.json'), 'w') as outld_fp:
            json.dump(jsonldcontent, outld_fp)
```

```python
graph = Graph()
graph.parse(os.path.join(data_dir, 'isa-rdf.json'))
```

```python
print(f"Graph g has {len(graph)} statements.")
```

```python
# Write turtle file
rdf_path=os.path.join(data_dir, "isa-rdf-" + ontology + "v3.ttl")
with open(rdf_path, 'w') as rdf_file:
        rdf_file.write(graph.serialize(format='turtle'))
```
